# **STEP 4:Feature Engineering (Date & Time Processing)**

```python
import pandas as pd
df=pd.read_csv("/content/traffic_congestion_dataset.csv")
df.columns = df.columns.str.strip()
df.columns
```

```python
datetime_col = None
for col in df.columns:
    if 'time' in col.lower() or 'date' in col.lower():
        datetime_col = col
        break

print("Detected datetime column:", datetime_col)
```

```python
datetime_col = 'Timestamp'
df[datetime_col] = pd.to_datetime(df[datetime_col], errors='coerce')
df = df.dropna(subset=[datetime_col]).copy()

df['hour'] = df[datetime_col].dt.hour
df['day'] = df[datetime_col].dt.day_name()
df['month'] = df[datetime_col].dt.month
df['year'] = df[datetime_col].dt.year
df['day_of_week_num'] = df[datetime_col].dt.weekday
df['is_weekend'] = df['day_of_week_num'].apply(lambda x: 1 if x >= 5 else 0)

df['peak_hour'] = df['hour'].apply(lambda x: 1 if (7 <= x <= 10 or 17 <= x <= 20) else 0)
```

# **STEP 5: Exploratory Data Analysis (EDA) & Visualization**

```python
import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(10,5))
sns.lineplot(x='hour', y="Vehicle_Count", data=df,color='navy')
plt.title('Traffic vs Hour')
plt.show()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
plt.figure(figsize=(10,5))
sns.barplot(x='day', y="Vehicle_Count", data=df,color='seagreen')
plt.xticks(rotation=45)
plt.title('Traffic vs Day')
plt.show()
```

```python
plt.figure(figsize=(8,5))
sns.histplot(df["Vehicle_Count"], bins=30,color='teal')
plt.title('Traffic Distribution')
plt.show()
```

```python
vehicle_cols = ['Car_Count', 'Bus_Count', 'Truck_Count', 'Bike_Count']

df[vehicle_cols].mean().plot(kind='bar', color='steelblue')

plt.title('Average Traffic by Vehicle Type')
plt.ylabel('Average Count')
plt.xticks(rotation=0)
plt.show()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
plt.figure(figsize=(8,5))
sns.histplot(df['Average_Speed'], bins=30, kde=True,color='slategray')

plt.title('Average Speed Distribution')
plt.xlabel('Speed')
plt.show()
```

```python
plt.figure(figsize=(8,5))
sns.lineplot(x='Travel_Time_Index', y='Vehicle_Count', data=df,color='navy')

plt.title('Traffic vs Travel Time Index')
plt.xlabel('Travel Time Index')
plt.ylabel('Vehicle Count')
plt.show()
```
